# MISO Neural Network — Multiple Inputs, Single Output

A MISO network takes several inputs and combines them into one output using a weighted sum (dot product): `prediction = input_1 × weight_1 + input_2 × weight_2 + ...`.

Local copy based on the [MISO Colab notebook](https://colab.research.google.com/drive/16SQuyfxPcXofLUkFmaJyGjBm36REYLwy?usp=sharing).

## The Scenario: Neo's Weather Shop

Temperature alone does not tell the whole story. Three things affect iced-coffee sales:

- Temperature in °C
- Footfall, measured in hundreds of people
- Ad spend in rand

All three inputs combine into one prediction: iced-coffee sales. That is MISO: many inputs, one output.

```python
# Multiply each input by its matching weight, then add the results
def w_sum(inputs, weights):
    assert len(inputs) == len(weights)
    total = 0
    for i in range(len(inputs)):
        total += inputs[i] * weights[i]
    return total

def neural_network(inputs, weights):
    prediction = w_sum(inputs, weights)
    return prediction
```

```python
# Weights: [temperature, footfall, ad spend]
weights = [2.0, 1.5, 0.3]

# Today's inputs: 25 C, 400 people (4 hundreds), and R10 ad spend
today_inputs = [25, 6, 10]
pred = neural_network(today_inputs, weights)
print(f"Inputs (temp, footfall, ad_spend): {today_inputs}")
print(f"Predicted iced coffees sold: {pred}")
```

### Expected result

The network multiplies each input by its matching weight and adds the results: `(25 × 2.0) + (4 × 1.5) + (10 × 0.3) = 50 + 6 + 3 = 59`. The prediction is **59.0 iced coffees**. These weights are hand-set for demonstration; they are not learned from actual sales.

```python
# A week's daily inputs: [temperature, footfall in hundreds, ad spend in rand]
week_inputs = [
    [15, 2, 5],
    [18, 3, 0],
    [20, 5, 10],
    [25, 4, 10],
    [30, 6, 20],
    [22, 3, 5],
    [17, 2, 0],
]
for day, inputs in enumerate(week_inputs, start=1):
    pred = neural_network(inputs, weights)
    print(f"Day {day}: inputs={inputs} -> predicted {pred} iced coffees")
```

### Expected result

Each prediction is the weighted sum of that day's inputs. Day 5, `[30, 6, 20]`, gives the highest prediction: `30 × 2.0 + 6 × 1.5 + 20 × 0.3 = 75.0`.

## The NumPy version

A weighted sum is a dot product. NumPy's `.dot()` performs the same multiplications and addition as our Python loop.

```python
import numpy as np
weights_np = np.array([2.0, 1.5, 0.3])

def neural_network_np(inputs, weights):
    prediction = inputs.dot(weights)
    return prediction

today_inputs_np = np.array([25, 6, 10])
pred_np = neural_network_np(today_inputs_np, weights_np)
print(f"Inputs (temp, footfall, ad_spend): {today_inputs_np}")
print(f"Predicted iced coffees sold (NumPy): {pred_np}")
```
