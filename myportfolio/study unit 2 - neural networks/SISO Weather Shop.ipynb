# SISO Neural Network — Single Input, Single Output

A SISO network answers the question: *Given one number, what single number should I predict?*

## The Scenario: Neo's Weather Shop

Neo runs a small drinks stand. She has noticed that on hot days she sells more iced coffee. This tiny neural network takes the temperature (°C) and predicts how many iced coffees she will sell.

The network computes `prediction = input × weight`. The weight is fixed at 2.5 so we can focus on how the network makes a prediction, not how it learns.

```python
# The network itself — this is the entire neural network
weight = 2.5

def neural_network(input, weight):
    prediction = input * weight
    return prediction
```

```python
# A single day: it is 20 degrees C today
temperature_today = 20
pred = neural_network(temperature_today, weight)
print(f"Temperature: {temperature_today} C")
print(f"Predicted iced coffees sold: {pred}")
```

### Expected result

With `weight = 2.5` and a temperature of `20`, the network multiplies them: `20 * 2.5 = 50`. You should see a prediction of **50.0 iced coffees**. The relationship is linear: each extra degree adds exactly 2.5 to the prediction.

```python
# Run the network across a whole week of temperatures
week_temperatures = [15, 18, 20, 25, 30, 22, 17]
for day, temp in enumerate(week_temperatures, start=1):
    pred = neural_network(temp, weight)
    print(f"Day {day}: {temp} C -> predicted {pred} iced coffees")
```

### Expected result

Each prediction is the day's temperature multiplied by `2.5`. For example, day 5 at 30°C predicts 75.0 coffees. With a positive weight, the hottest day gives the highest prediction.

## The NumPy version

For a single number, the NumPy version looks almost identical to plain Python. NumPy becomes more useful when we work with lists and matrices, as in MISO, SIMO, and MIMO networks.

```python
import numpy as np
weight_np = np.array(5)

def neural_network_np(input, weight):
    prediction = input * weight
    return prediction

temperature_today = 20
pred_np = neural_network_np(temperature_today, weight_np)
print(f"Temperature: {temperature_today} C")
print(f"Predicted iced coffees sold (NumPy): {pred_np}")
```

## Where to experiment

- Change `weight` (try `5` or `-1`) and rerun the cells to see how it changes the prediction. A negative weight could represent hot-soup sales, where colder days increase demand.
- Change `temperature_today` or values in `week_temperatures` (try `40` or `0`).
- Change `weight_np` as well and check that the NumPy prediction still matches the plain-Python one.

Because SISO has one input and one weight, there is one input value and one scaling factor to explore.
